```python
a=int(input("enter first number:"))
b=int(input("enter second number:"))
choice=int(input("enter your choice:\n1.Addition\n2.substraction\n3.multiplication\n4.division\n"))
if(choice==1):
    print("sum=",a+b)
elif(choice==2):
    print("difference",a-b)
elif(choice==3):
    print("multiplication",a*b)
elif(choice==4):
    print("division",a/b)
else:
    print("invalid input!")
    
```
