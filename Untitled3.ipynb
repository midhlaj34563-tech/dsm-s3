```python
import matplotlib.pyplot as plt

mark = [24,30,40,29,55]
sub = ["chem","phy","maths","bio","iot"]
plt.title("plot")
plt.plot(mark,sub,'r:o')
plt.xlabel("mark")
plt.ylabel("sub")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [42, 30, 30, 23, 24]

x = np.array([42, 32, 30, 21, 23])
y = np.array([42, 30, 30, 23, 24])

plt.plot(x, y, marker='*', color='red', linewidth=3, label="Long Line")

plt.plot(
    y,
   x
)

plt.title("Marks Plot")
plt.xlabel("Subjects")
plt.ylabel("Marks")


plt.show()
```
