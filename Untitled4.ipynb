```python
import matplotlib.pyplot as plt

groups = ["G1", "G2", "G3", "G4", "G5"]
men = [22, 30, 35, 35, 26]
women = [25, 32, 30, 35, 29]

x = [0, 1, 2, 3, 4]

plt.bar(x, men, width=0.4, label="Men", color="red")
plt.bar([i + 0.4 for i in x], women, width=0.4, label="Women", color="orange")

plt.xticks([i + 0.2 for i in x], groups)
plt.xlabel("Groups")
plt.ylabel("Scores")
plt.title("Scores by Group and Gender")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]
y2 = [1, 3, 5, 7]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.title("Two Lines")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

language = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]

plt.pie(popularity, labels=language, autopct="%1.1f%%")
plt.title("Programming Language Popularity")
plt.show()
```

```python
 import matplotlib.pyplot as plt

language = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]

x = range(len(language))

plt.scatter(x, popularity)
plt.xticks(x, language)
plt.xlabel("Language")
plt.ylabel("Popularity")
plt.title("Programming Language Popularity")
plt.show()
```

```python
import matplotlib.pyplot as plt

language = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]

plt.barh(language, popularity)
plt.xlabel("Popularity")
plt.title("Programming Language Popularity")
plt.grid(True)

plt.show()
```
